# Q.2

```python
import numpy as np
from scipy import stats
A = np.array([
    [5, 2, 8],
    [1, 9, 3],
    [6, 4, 7]
])
print("Original array:\n", A)
print("Descending entire array:")
print(np.sort(A.flatten())[::-1])
print("Row-wise descending:")
print(-np.sort(-A, axis=1))
print("Column-wise descending:")
print(-np.sort(-A, axis=0))
print("Mean:", np.mean(A))
print("Median:", np.median(A))
print("Mode:", stats.mode(A.flatten(), keepdims=True).mode[0])
print("Variance:", np.var(A))
print("Standard deviation:", np.std(A))
```

```python
import numpy as np
numbers = list(map(int, input("Enter 9 integers: ").split()))
A = np.array(numbers).reshape(3, 3)
print("Array:")
print(A)
print("Determinant:", np.linalg.det(A))
print("Inverse:")
print(np.linalg.inv(A))
values, vectors = np.linalg.eig(A)
print("Eigenvalues:")
print(values)
print("Eigenvectors:")
print(vectors)
```

# Q.7

```python
import numpy as np
n = int(input("Enter the size of the checkerboard: "))
board = np.zeros((n, n), dtype=int)
for i in range(n):
    for j in range(n):
        if (i + j) % 2 == 1:
            board[i][j] = 1
print(board)
```

# Q.9

```python
import numpy as np
rainfall = np.array([
    [12, 15, 18, 10, 20, 14, 16],
    [8, 10, 12, 9, 15, 11, 13],
    [20, 18, 22, 25, 24, 21, 23],
    [5, 7, 6, 8, 9, 10, 7],
    [14, 16, 15, 18, 17, 19, 20]
])
total_city = np.sum(rainfall, axis=1)
average_day = np.mean(rainfall, axis=0)
print("Total rainfall for each city:")
print(total_city)
print("Average rainfall for each day:")
print(average_day)
highest_city = np.argmax(total_city) + 1
print("City with highest rainfall: City", highest_city)
print("Rainfall values greater than 20:")
print(rainfall[rainfall > 20])
print("Maximum rainfall:", np.max(rainfall))
print("Minimum rainfall:", np.min(rainfall))
```

# Q.10

```python
import random
number = ''.join(random.sample('0123456789', 4))
print("Welcome to the Cows and Bulls Game!")
attempts = 0
while True:
    guess = input("Enter a 4-digit number: ")
    if len(guess) != 4 or not guess.isdigit():
        print("Please enter exactly 4 digits.")
        continue
    attempts += 1
    cows = 0
    bulls = 0
    for i in range(4):
        if guess[i] == number[i]:
            cows += 1
        elif guess[i] in number:
            bulls += 1
    print(cows, "cows,", bulls, "bulls")
    if cows == 4:
        print("Congratulations! You guessed the number.")
        print("Total attempts:", attempts)
        break
```
